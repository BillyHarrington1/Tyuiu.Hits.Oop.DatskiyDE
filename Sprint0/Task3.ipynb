<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс BankAccount в C#, который будет представлять
информацию об учетных записях в банке. На основе этого класса разработать 2-3
производных класса, демонстрирующих принципы наследования и полиморфизма.
В каждом из классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Требования к базовому классу BankAccount:
• Атрибуты: Номер счета (AccountNumber), Баланс (Balance), Тип счета
(AccountType).

• Методы:

o GetInfo(): метод для получения информации о счете в виде строки.

o Deposit(decimal amount): метод для внесения денег на счет.

o Withdraw(decimal amount): метод для снятия денег со счета.

Требования к производным классам:
1. Сберегательный счет (SavingsAccount): Должен содержать дополнительные
атрибуты, такие как Процентная ставка (InterestRate). Метод Deposit() должен
быть переопределен для добавления процентов к сумме вклада при
внесении денег на счет.
2. Текущий счет (CheckingAccount): Должен содержать дополнительные
атрибуты, такие как Лимит овердрафта (OverdraftLimit).
Метод Withdraw() должен быть переопределен для проверки и применения
лимита овердрафта при снятии денег со счета.
3. Инвестиционный счет (InvestmentAccount) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Список активов
(AssetsList). Метод GetInfo() должен быть переопределен для включения
информации о списках активов в описании счета.

```
BankAccount savings = new SavingsAccount("408-17-810-7-0000-0000007", 10000, "Сберегательный", 4.5m);
BankAccount checking = new CheckingAccount("408-17-810-8-0000-0000008", 5000, "Текущий", 2000);
BankAccount investment = new InvestmentAccount("408-17-810-9-0000-0000009", 50000, "Инвестиционный", ["Акции Сбербанка", "ОФЗ-26238", "Золото"]);

savings.Deposit(2000); //полиморфный метод с расчетом процентов
Console.WriteLine(savings.GetInfo());
Console.WriteLine();

checking.Withdraw(6500); //полиморфный метод с проверкой ухода в минус
Console.WriteLine(checking.GetInfo());
Console.WriteLine();

Console.WriteLine(investment.GetInfo());


public abstract class BankAccount
{
    private string _accountNumber;
    private decimal _balance;
    private string _accountType;

    public string AccountNumber
    {
        get => _accountNumber;
        set { if (!string.IsNullOrWhiteSpace(value)) _accountNumber = value; }
    }

    public decimal Balance
    {
        get => _balance;
        protected set => _balance = value;
    }

    public string AccountType
    {
        get => _accountType;
        set { if (!string.IsNullOrWhiteSpace(value)) _accountType = value; }
    }

    protected BankAccount(string accountNumber, decimal initialBalance, string accountType)
    {
        AccountNumber = accountNumber;
        Balance = initialBalance;
        AccountType = accountType;
    }

    public virtual void Deposit(decimal amount) 
    {
        if (amount > 0)
        {
            Balance += amount;
        }
    }

    public virtual void Withdraw(decimal amount)
    {
        if (amount > 0 && Balance >= amount)
        {
            Balance -= amount;
        }
    }

    public virtual string GetInfo()
    {
        return $"Счёт: {AccountNumber} | Тип: {AccountType} | Баланс: {Balance:C}";
    }
}

public class SavingsAccount : BankAccount //сберег.счет - класс и наследование(наш сбер.счет к учетной записи), 71 строка 
{
    private decimal _interestRate; //закрытый 

    public decimal InterestRate //атрибут - Процентная ставка (с геттером и сеттером)
    {
        get => _interestRate;
        set { if (value >= 0) _interestRate = value; } //не - ли?
    }

    public SavingsAccount(string accountNumber, decimal initialBalance, string accountType, decimal interestRate) 
        : base(accountNumber, initialBalance, accountType) //конструктор базового класса
    {
        InterestRate = interestRate;
    }

    public override void Deposit(decimal amount) //Полиморфизм - Переопределение(override) метода внесения денег с расчетом процентов
    {
        if (amount > 0)
        {
            decimal bonus = amount * (InterestRate / 100);
            base.Deposit(amount + bonus);
        }
    }

    public override string GetInfo() //Переопределение метода информации о счете
    {
        return base.GetInfo() + $" | Ставка: {InterestRate}%";
    }
}

public class CheckingAccount : BankAccount // тек.счет - класс и наследование
{
    private decimal _overdraftLimit;

    public decimal OverdraftLimit
    {
        get => _overdraftLimit;
        set { if (value >= 0) _overdraftLimit = value; }
    }

    public CheckingAccount(string accountNumber, decimal initialBalance, string accountType, decimal overdraftLimit) 
        : base(accountNumber, initialBalance, accountType) //конструктор
    {
        OverdraftLimit = overdraftLimit;
    }

    public override void Withdraw(decimal amount) ///Переопределение метода снятия денег с учетом овердрафта(вывод средств через сверхлимит)
    {
        if (amount > 0 && (Balance + OverdraftLimit) >= amount) 
        {
            Balance -= amount;
        }
    }

    public override string GetInfo() //Переопределение метода информации о текущего счета
    {
        return base.GetInfo() + $" | Лимит овердрафта: {OverdraftLimit:C}";
    }
}

public class InvestmentAccount : BankAccount //инв.счет - класс и наследование
{
    private List<string> _assetsList = new();

    public List<string> AssetsList //атрибут - список активов 
    {
        get => _assetsList;
        set { if (value != null) _assetsList = value; }
    }

    public InvestmentAccount(string accountNumber, decimal initialBalance, string accountType, List<string> assetsList) 
        : base(accountNumber, initialBalance, accountType) //конструктор 
    {
        AssetsList = assetsList;
    }

    public override string GetInfo() //Переопределение метода информации для вывода активов:
    {
        string assets = AssetsList.Count > 0 ? string.Join(", ", AssetsList) : "Нет активов";
        return base.GetInfo() + $" | Активы: [{assets}]";
    }
}
```
